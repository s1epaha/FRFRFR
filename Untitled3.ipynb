```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

# Загрузка данных с правильным разделителем
url = 'https://drive.google.com/uc?export=download&id=1R-bDTTCsLjzy9NZG3bkzzeNnFyBuupGK'
data = pd.read_csv(url, sep=';')

# Переименуем колонки для удобства
data.columns = ['forecast', 'station_a', 'station_b', 'station_c', 'station_d']

# Проверим данные
print("Первые строки данных:")
print(data.head())

# Создание графиков ошибок для каждой СЭС
plt.figure(figsize=(20, 12))

# Список станций
stations = ['station_a', 'station_b', 'station_c', 'station_d']
station_names = ['A', 'B', 'C', 'D']

# Вычисление ошибок и RMSE для каждой СЭС
rmse_values = {}

for i, (station, name) in enumerate(zip(stations, station_names), 1):
    # Вычисление ошибки прогноза
    error = data['forecast'] - data[station]

    # Расчет RMSE
    rmse = np.sqrt(mean_squared_error(data[station], data['forecast']))
    rmse_values[name] = rmse

    # Построение графика ошибок
    plt.subplot(2, 2, i)
    plt.plot(error, 'r')
    plt.title(f'Ошибка прогноза для СЭС {name} (RMSE: {rmse:.2f})')
    plt.xlabel('Время, ч')
    plt.ylabel('Ошибка, МВт')
    plt.grid(True)

plt.tight_layout()
plt.savefig('solar_forecast_errors.png')
plt.show()

# Вывод RMSE для всех СЭС
print("\nRMSE для каждой СЭС:")
for name, rmse in rmse_values.items():
    print(f"СЭС {name}: {rmse:.2f} МВт")

# Средний RMSE по всем СЭС
avg_rmse = np.mean(list(rmse_values.values()))
print(f"\nСредний RMSE по всем СЭС: {avg_rmse:.2f} МВт")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

# Загрузка данных с указанием правильного разделителя
url = 'https://drive.google.com/uc?export=download&id=1R-bDTTCsLjzy9NZG3bkzzeNnFyBuupGK'
data = pd.read_csv(url, sep=';')

# Проверим названия столбцов
print("Доступные столбцы:", data.columns.tolist())

# Переименуем столбцы для удобства (если нужно)
data.columns = ['forecast', 'station_a', 'station_b', 'station_c', 'station_d']

# Создадим график
plt.figure(figsize=(15, 10))

# Список станций и их обозначений
stations = {
    'station_a': 'A',
    'station_b': 'B',
    'station_c': 'C',
    'station_d': 'D'
}

# Рассчитаем RMSE для каждой станции
rmse_results = {}
for station, name in stations.items():
    # Вычисляем ошибки прогноза
    errors = data['forecast'] - data[station]

    # Рассчитываем RMSE
    rmse = np.sqrt(mean_squared_error(data[station], data['forecast']))
    rmse_results[name] = rmse

    # Строим график ошибок
    plt.plot(errors, label=f'СЭС {name} (RMSE={rmse:.2f})')

# Настроим график
plt.title('Ошибки прогноза выработки для СЭС A-D')
plt.xlabel('Время, часы')
plt.ylabel('Ошибка прогноза, МВт')
plt.legend()
plt.grid(True)
plt.tight_layout()

# Сохраним график
plt.savefig('forecast_errors.png')
plt.show()

# Выведем результаты RMSE
print("\nРезультаты оценки точности прогноза (RMSE):")
for name, rmse in rmse_results.items():
    print(f"СЭС {name}: {rmse:.2f} МВт")

# Среднее значение RMSE
mean_rmse = np.mean(list(rmse_results.values()))
print(f"\nСреднее RMSE по всем станциям: {mean_rmse:.2f} МВт")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

# Загрузка данных (разделитель — точка с запятой)
url = 'https://drive.google.com/uc?export=download&id=1R-bDTTCsLjzy9NZG3bkzzeNnFyBuupGK'
data = pd.read_csv(url, sep=';')

# Проверяем названия столбцов (может потребоваться переименование)
print("Исходные столбцы:", data.columns.tolist())

if len(data.columns) == 1:
    data = data.iloc[:, 0].str.split(';', expand=True)
    data.columns = ['Forecast', 'Station_A', 'Station_B', 'Station_C', 'Station_D']

# Преобразуем данные в числа (на случай, если они хранятся как строки)
for col in data.columns:
    data[col] = pd.to_numeric(data[col], errors='coerce')  # 'coerce' превращает нечисловые значения в NaN

# Проверяем пропуски (NaN)
print("\nПропущенные значения в данных:")
print(data.isnull().sum())
data_clean = data.dropna()
if len(data_clean) < 2:
    raise ValueError("После удаления NaN осталось слишком мало данных для анализа!")

# Создаем график
plt.figure(figsize=(50, 50))

# Словарь станций
stations = {
    'station_a': 'A',
    'station_b': 'B',
    'station_c': 'C',
    'station_d': 'D'
}

# Рассчитываем RMSE для каждой станции
rmse_results = {}

for station, name in stations.items():
    # Вычисляем ошибки прогноза
    errors = data_clean['Forecast'] - data_clean[station]

    # Рассчитываем RMSE (игнорируя NaN, если они остались)
    rmse = np.sqrt(mean_squared_error(
        data_clean[station],
        data_clean['forecast']
    ))
    rmse_results[name] = rmse

    # Строим график ошибок
    plt.plot(errors, label=f'СЭС {name} (RMSE={rmse:.2f})')

# Настраиваем график
plt.title('Ошибки прогноза выработки для СЭС A-D')
plt.xlabel('Время, часы')
plt.ylabel('Ошибка прогноза, МВт')
plt.legend()
plt.grid(True)
plt.tight_layout()

# Сохраняем график
plt.savefig('forecast_errors.png')
plt.show()

# Выводим RMSE для каждой станции
print("\nРезультаты оценки точности прогноза (RMSE):")
for name, rmse in rmse_results.items():
    print(f"СЭС {name}: {rmse:.2f} МВт")

# Среднее RMSE
mean_rmse = np.mean(list(rmse_results.values()))
print(f"\nСреднее RMSE по всем станциям: {mean_rmse:.2f} МВт")
```
